```python
#lets import
import pandas as pd
```

```python
#lets create a simple data frame
#To create a data frame we use a dictionary
my_cars={
    "car":['BMW','Ferari','Mazda'],
    "prices":[190000,200000,10000]
}

df_cars=pd.DataFrame(my_cars)
print(df_cars)
```

**Series**

A pandas series is like a column in a table

It is a one dimension array holding data of any type



```python
#simple series
a=[1,2,3,4]

#add the column using pandas
df_a=pd.Series(a)

#display the series
print(df_a)
```

**Labels**

If nothing else is specified the values are labled with their index numbers.

Index starts at '0'

This labels(index) Can be used to access a specified value.

*N/B*: This only works on a series not a dataframe. When working with dataframes use iloc[] instead

```python
#lets access value "4" in our df_a
print(df_a[3])
```

```python
#lets access information on "Ferari" in our df_cars------Dataframe
print(df_cars.iloc[2])
```

**Create Labels**

With the index argument we can label our own labels

```python
#create a series
a=[100,200,300]

#now lets add our label
myvar=pd.Series(a, index=['Juja','Thika','Nyeri'])

#display
print(myvar)
```

```python
#accessing items of the series above with user defined labeld
print(myvar['Thika'])
```

**Key/Value Objects as a series**

Using a dictionary when creating a series

*Note:* The keys of the dictionaries become the labels


```python
#lets create a dictionary to pass to our series

calories={"day1":420,"day2":300,"day3":400, "day4":350,"day5":388}

#add the dictionary to a pandas series
df_calories=pd.Series(calories)

#display
print(df_calories)
```

```python
#we can select only what we need to be in our series using the index
df_new_calories=pd.Series(calories, index=['day1','day3'])

#display
print(df_new_calories)
```

**DataFrames**

Data sets in pandas are usually multi dimensional tables called DataFrames

*NOTE*: Series is like a column while a dataframe is the whole table

```python
#Create a data frame from two series EG Students and there fee balance

student={
    "Name":["Allan","Jaba","Mark"],
    "Fee_Balance":[10000, 5799,11500]
}

#load the dataframe
df_students=pd.DataFrame(student, index=['BIT','BBC','BMIT'])

#display
print(df_students)
```

**More on DataFrames...**

Is a two dimension data structure with rows and columns


```python
#create dataframe with two series
data={
    "calories":[420,380,390],
    "duration(mins)":[50,40,45]
}

df_exercise=pd.DataFrame(data)

#display the data frame
print(df_exercise)
```

**Locate Row**

Dataframe as shown from the results above is a table with rows and columns

Pandas use the **loc** attribute to return one or more specified row(s)

```python
#refer to the row index
print(df_exercise.loc[1])
```

```python
#this example below wil return a pandas series(1 dimension remember)
print(df_exercise.loc[[0,2]])  #we are using a list of indexes
```

**Elements in a Data Frame**

When someone asks: “How many elements does this DataFrame have?”

Refers to the total number of data points(cells), not just rows
In Pandas:

Number of rows → df.shape[0]

Number of columns → df.shape[1]

Total elements (rows × columns) → df.size

```python
import pandas as pd


company = {
    "Emp_Name": ['Allan', 'Jaba', 'Mark'],
    "Languages": ['Python', 'JavaScript', 'Bash'],
    "Salary": [40000, 86999, 150000]
}

#load the dataframe
df_company=pd.DataFrame(company)

#display
print(df_company)
```

```python
#now display the number of rows and columns
df_company.shape
```

```python
#display rows only
print(f'ROW: {df_company.shape[0]}')
```

```python
#display columns only
print(f'COLUMNS: {df_company.shape[1]}')
```

```python
#get the total elements inthe dataframe
print(f'Total Elements inthe dataframe: {df_company.size}')
```

**Getting Max() and Min()**

Max() method is used to get the highest value in a column. Check to the column data type before using this

Min() method is used to identify the lowest value in a column

```python
df_company['Salary'].max()  #gets the highest paid employee
```

```python
df_company['Salary'].min()   #gets the lowest paid employee
```

```python
# We can use sort_values method to sort the columns by the one with highest salary
df_company.sort_values(by='Salary', ascending=False)
```

```python
#Let set the first column in our above dataframe to be our index
#Specify the column name followed by the inplace attribute to make the changes permanent
df_company.set_index('Emp_Name', inplace=True)
```

```python
#display
df_company
```

```python
df_company.loc['Mark'] #If we want now to access a value
```

```python
#Lets use a condition check the which row has the salary == 40000
df_company[df_company['Salary']==40000]
```

```python
#Here the first argument inside the loc is the index, followed by the column we want to check
df_company.loc['Mark', 'Salary']
```

```python

#who has a salary of 40000, lets assume we are iterating through a large column
names=['Mark','Jaba','Allan']

#write a for loop to loop through the column index and an if condition inside the for loop to check if someone is earning 40000
for name in names:
  #here we pass in the index and the row we want to check
  if df_company.loc[name, 'Salary']==40000:
    print(f'correct answer: {name}')
  else:
    print(f'{name}: not it')
  print()
```

```python
#what is the highest possible length of words in our index
max_lenght=max(df_company.index.str.len())

print(max_lenght)
```

**MODE**

In a column we may want to identify the most frequent value

We use the mode() method for this

```python
data = {
    "Emp_Name": ["Allan", "Jaba", "Mark", "Allan"],
    "Salary": [40000, 86999, 150000, 40000]
}

df_startup = pd.DataFrame(data)
```

```python
#check salary column
df_startup['Salary'].mode()
```

```python
#check the employee column
df_startup['Emp_Name'].mode()
```
