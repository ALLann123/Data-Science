# **Manipulating Data(Applying Functions)**

Lets manipulate individual columns

```python
#we can double the age column
df['age']=df['age']*2
```

```python
#display
df
```

```python
def myfunction(x):
  #checks if value is divisible by 3 then squares the value
  if x % 3==0:
    return x**2
  else:
    return x//2
```

```python
df['age'].apply(myfunction)
```

```python
def myfunction2(x):
  if x.endswith('s'):
    return 'Fired'
  else:
    return 'Safe'
```

```python
#lets apply the function- Make a new column of this expression
df['Reviews']=df['job'].apply(myfunction2)

#display
df
```

```python
#if the function is simple we can use a lambda function
df['age'].apply(lambda x:x**2 if x % 3==0 else x // 2)
```

```python
#create a new column called summary of the other columns
df['summary']=df.apply(lambda row:f"Age: {row['age']}, Job: {row['job']}, review: {row['Reviews']}", axis=1)

#display
df
```

```python
#How to drop columns
df=df.drop('summary', axis=1)
```

```python
#view
df
```

```python
df.set_index('name', inplace=True)
```

# **Data Cleaning**

Checking and removing null values, incorrect column formats and wrong values that affect the data in general.

```python
#lets add a null value to our age column
df.at['Allan','age']=float('nan')
```

```python
#lets set something else to null
df.at['Jaba','job']=float('nan')
```

```python
#display
df
```

```python
#get the data summary to know the null values
df.info()
```

```python
df.isnull().sum()  #get the column and the number of missing values
```

```python
#---1.) We can drop the row with null values
df.dropna()
```

```python
#---2) We can fill the null value
df.fillna(0)
```

```python
#----3) Drop only rows with null values. Drops certain rows on specified columns
df[df['age'].notna()]
```

# **Iterating**

Using a For loop to go over our individual columns.



```python
#we iterate through individual rows
for i, row in df.iterrows():
  print(i)
```

```python
#iterate on the rows
for i, col in df.items():
  print(i)
```

**Filtering and Querying**

```python
df['age']<29
```

```python
#multiple conditions Using the and operator(*)
df[(df['age']<29) & (df['job'].notna())]
```

```python
#we can use a logical or(|)
df[(df['age']<29) | (df['job'].notna())]
```

```python
df
```

```python
import datetime as dt

df['birthday'] = df['age'].apply(
    lambda x: dt.datetime.now() - dt.timedelta(days=365 * x)
)

df
```

```python
#we can check for a list
ages=[23,24]

#we can loop through a list using the isin() method
df[df['age'].isin(ages)]
```

```python
#The QUery method-- It is fast as its build on numpy and can be used when with large data sets
df.query('age==25')
```
