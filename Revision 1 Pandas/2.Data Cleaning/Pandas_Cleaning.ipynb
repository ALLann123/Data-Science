**Pandas Cleaning**

Refers to fixing bad data in your data set.

*Bad Data Could be:*
1. Empty cells
2. Data in wrong format
3. Wrong data
4. Duplicates



```python
import pandas as pd
```

```python
#lets load our csv
df_data=pd.read_csv('sample_data/data.csv')

#display
df_data.head()
```

```python
df_data.shape[0]
```

```python
#lets add a messy date column to the dataframe=---This will be perfect for cleaning
import numpy as np
import random


# assume your DataFrame is called df with 169 rows
n = len(df_data)   # should be 169

# generate messy dates
bad_dates = []
for i in range(n):
    choice = random.choice([1, 2, 3, 4, 5])
    if choice == 1:
        bad_dates.append("2023-01-{:02d}".format((i % 28) + 1))  # YYYY-MM-DD
    elif choice == 2:
        bad_dates.append("01/{:02d}/2023".format((i % 28) + 1))  # MM/DD/YYYY
    elif choice == 3:
        bad_dates.append("202301{:02d}".format((i % 28) + 1))    # YYYYMMDD
    elif choice == 4:
        bad_dates.append("Jan {:02d}, 2023".format((i % 28) + 1)) # text format
    else:
        bad_dates.append(np.nan)  # missing value

# add as new column
df_data["Date"] = bad_dates

# preview
print(df_data.head(10))
```

```python
#display the dataframe
df_data
```

```python
#lets see how many null values are in each column. Use the isnull() method to identify null values
df_data.isnull().sum()   #we know the date column has plenty of null values
```

**Empty Cells**

Empty cells can potentially give you wrong result when you analyze data


**Remove Rows**

One way to deal with empty cells is to remove rows that contain empty cells

*NOTE*: This is ok because data sets can be very big, and removing a few rows will not have a big impact on the result.

**dropna()** method can perform this on our dataframe


N/B: dropna() method returns a new DataFrame and will not change the original. If we want to change the original dataframe add the **inplace=True** argument.

```python
#lets drop all null values in our loaded data frame
new_df=df_data.dropna()

#display
new_df
```

```python
#check if there are any null values in any column
new_df.isnull().sum()
```

**Replace Empty Values**

Another way of dealing with empty cells is to insert a new value instead.

This prevents deleting entire rows just because of some empty cells

**fillna()** method allows us to replace empty cells with a value we choose.

```python
df_replace=df_data   #make a copy of our dataframe

df_replace.head(4)
```

```python
#check the null values inthe Calories column
df_replace['Calories'].isnull().sum()
```

```python
#lets replace the Calories column Null values with the number 130

df_replace.fillna({"Calories":130}, inplace=True)
```

```python
#lets check the five null rows have the value 130
df_replace[df_replace["Calories"]==130]   #they should be five
```

**Replace Using Mean, Median, or Mode**

A common way to replace empty cells is to calculate the mean, median or mode value of the column.

Pandas uses the **Mean()**, **Median()**, **Mode()** methods to calculate the respoective values for a specified column

**a)Mean()**

 the arithmetic average of a group of numbers, calculated by summing all values and dividing by the total number of values

 We use the .mean() method in pandas


```python
#make a copy of the data frame
df_mean_replace=df_data
```

```python
#Calculate the MEAN() of the column with null values
x=df_mean_replace['Calories'].mean().round(2)

#display the mean
print(x)
```

```python
#now use the calculated mean above stored in 'x' to replace the null values
df_mean_replace.fillna({'Calories':x}, inplace=True)
```

```python
#now lets check the 5 empty rows inthe Calories column if filled by the mean
df_mean_replace[df_mean_replace['Calories']==375.79]
```

**b)MEDIAN()**

Refers to the middle value in a data set that has been ordered from the least to greatest.  Represents the 50th percentile meaning that half the data points are below it and half are above it.


```python
#load the dataset
df_median_replace=pd.read_csv('sample_data/data.csv')
```

```python
#check null values in calories
df_median_replace.isnull().sum()
```

```python
#get the median
x=df_median_replace['Calories'].median()

#display
print(x)
```

```python
#Now add the calculated median to all the values
df_median_replace.fillna({'Calories':x},inplace=True)
```

```python
#check if the 5 rows have the median value inthe calories column
df_median_replace[df_median_replace['Calories']==318.6]
```

**c)MODE()**

Refers to the value that appears most frequently in a data set.

```python
#load the dataset
df_mode_replace=pd.read_csv('sample_data/data.csv')
```

```python
#we know the calories column has the null 5 null values
#Lets calculate the mode and get the first element of te result
x=df_mode_replace['Calories'].mode()[0]

#display
print(x)
```

```python
#use the mode we got to replace the value
df_mode_replace.fillna({'Calories':x}, inplace=True)
```

```python
#check the 5 rows, Remember Mode is the value that reappears so the answer will be more than 5 rows
df_mode_replace[df_mode_replace['Calories']==300.0]
```

# **Cleaning Data of Wrong Format**

Cells with wrong data format make it difficult to analyze data.

To fix it you have two options:
1. Remove the rows
2. Convert the cells inthe column into the same format



**Convert Into a Correct Format**

In our data frame the date column has date formats not of the correct format and contains null values

**to_datetime()** method can convert the column values into dates

```python
df_data.head() #display the dataframe
```

```python
#lets convert all cells inthe "Date Column into dates" format='mixed' tells pandas to not assume and get all formats from the column
df_data['Date']=pd.to_datetime(df_data['Date'], format='mixed')
```

```python
#Check the datframe now
df_data
```

```python
#check the null values inthe columns
df_data.isnull().sum()
```

```python
#lets clean now the column by droping rows with null values from our dataframe
df_data.dropna(inplace=True)
```

```python
df_data.isnull().sum() #check if there are any null values
```

**Example: to_datetime() Method**


```python
#lets create our dataframe object(contains two series)
activity={
    "Date":["2023-01-01", "01/02/2023", "20230103", "Jan 04, 2023", None],
    "Activities":['Running','Skating','Gym Work', 'Football','Coding']
}

df_activities=pd.DataFrame(activity)

```

```python
#display our dataframe
df_activities
```

```python
#lets clean the messy date column in one go
df_activities['Date']=pd.to_datetime(df_activities['Date'], format='mixed')

#display the cleaned dataframe
df_activities
```

**NOTE**:
1. All formats inthe dataframe are unified to YYYY-MM-DD
2. The missing value becomes NaT(Not a Time) which is the datetime's version of NaN


# **Fixing Wrong Data**

**Wrong Data**

Does not have to be empty cells or wrong formats but it can just be wrong e.g someone registering "199" instead of "1.99"


In our dataframe we can say that a person working out more than 120 minutes isnt mostly realistic as majority of the data lies between 60 minutes to 100 minutes. They may have made an error.

```python
#Identify the different values inthe column and how many they are
df_data['Duration'].value_counts()
```

```python
#lets display all rows that are above 120 minutes
df_data[df_data['Duration']>120]
```

```python
#Lets get the median and set on these rows
y=df_data['Duration'].median()

#display
print(y)
```

```python
#now lets loop through all the values inthe duration column and set anything higher than 120 to the median value
for x in df_data.index:
  #we access the exact row using its index and the second argument is the column name
  if df_data.loc[x, "Duration"]>120:
    #lets set the value to 60 which is our calculated median
    df_data.loc[x, "Duration"]=y
```

```python
#check it now
df_data['Duration'].value_counts()
```

# **Removing Duplicates**

Duplicates are rows that have been registered more than one time.

To discover about duplicates we use the **duplicated()** method- returns Bools of each row.


```python
#lets look if our dataframe has duplicates
df_data.duplicated().sum()   #our dataframe has no duplicates
```

**Removing Duplicates**

To remove duplicates, use the drop_duplicates() method



```python
#create object
company={
    "Name":['Allan','Jaba','Rembo','Mark','Allan','Alvin'],
    'Languages':['Python','JS','SQL','Bash','Python','C/C++'],
    'Salary':[40000, 86999,30000,150000,40000,100000]
}

#load dataframe
df_company=pd.DataFrame(company)

#display
df_company
```

```python
#lets identify duplicates
df_company.duplicated().value_counts()
```

```python
#we got one duplicated row, drop it. Add inplace=True
df_company.drop_duplicates(inplace=True)
```

```python
#display new df
df_company
```
