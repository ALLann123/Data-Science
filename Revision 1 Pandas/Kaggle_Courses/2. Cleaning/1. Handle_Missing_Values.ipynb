# **Handling Missing Values**

Data Cleaning involves handling missing values, incorrect column formats and inconsistent data entry.

**1. Take a first look at the data**

This gives us an idead of whats going on with the data.

```python
#get the first 5 rows
nfl_data.head()  #we see missing values already
```

```python
col_list=list(nfl_data.columns)

for i in col_list:
  print(i)
```

```python
#take a few columns to work with
nfl_data=nfl_data[[
    "Date",
    "GameID",
    "Drive",
    "qtr",
    "down",
    "time",
    "TimeUnder",
    "TimeSecs",
    "PlayTimeDiff",
    "SideofField"
]]

#get 5 rows
nfl_data.head()
```

**2. How many missing data points do we have**

Identify how many missing values we have in each column

```python
#get total missing values per column
missing_values_count=nfl_data.isnull().sum()

#look at the first 10 columns
missing_values_count[0:10]
```

```python
nfl_data.shape
```

That seems like alot, lets find the percentage of the values in our dataset were missing to give us a better sense of the scale of this problem

```python
import numpy as np
```

```python
#how many total missing values do we have
total_cells=np.prod(nfl_data.shape) #use numpy to multiply rows and columns
total_missing=missing_values_count.sum() #count all the missing values inthe dataframe


#percentage of data that is  missing
percent_missing=(total_missing/total_cells)*100

print(f"{percent_missing}%") #get the percentage of data missing
```

# **Quick Fixes**

**1. Drop missing values**

When in a hurry or dont have time to figure out what values are missing, one option you have is to remove any rows or columns that contain missing values.

**NOTE**: Not reccomended when working on important projects.

We can drop missing values using **dropna()** method.

```python
#remove all rows that contain NA
nfl_data.dropna()
```

😂😂That removed all the rows in our data as they all contained null value.

Lets remove columns that have a null values instead.

```python
#remove columns with missing vals
columns_with_na_dropped=nfl_data.dropna(axis=1)

#display
columns_with_na_dropped.head()
```

```python
#confirm
columns_with_na_dropped.isnull().sum()
```

```python
#just how much data did we lose?
print("Columns in original dataset: ", nfl_data.shape[1])

print("Columns with na's dropped: ", columns_with_na_dropped.shape[1])
```

```python
col_num_one=nfl_data.shape[1]

col_num_two=columns_with_na_dropped.shape[1]

print(col_num_two)
print(col_num_one)

difference=col_num_one-col_num_two
difference
```

We lost abit of our data but we successfully removed all the NaN's from our data.

**2. Filling missing values automatically**

Try and fill missing values using **fillna()** method. We specify what we want the NaN values to be replaced with.

```python
#get small subset of the data
subset_nfl_data=nfl_data.loc[:,'EPA':'Season'].head()

#display
subset_nfl_data
```

```python
#replace all NA's with 0
subset_nfl_data.fillna(0)
```

Make it more savvy and replace missing values with whenever value comes directly after it in the same column.

*Note*: Makes more sense in datasets where observation has some logic order to them.



```python
#fill NA with value after it and when you get to end without value fill with 0
subset_nfl_data.fillna(method="bfill", axis=0).fillna(0)
```
