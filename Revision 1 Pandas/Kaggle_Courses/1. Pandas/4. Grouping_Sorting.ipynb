# **Grouping and Sorting**

groupby() operation, allows us to group our data and then do something specific to the group the data is in.

**Groupwise analysis**

We can replicate what value_counts() byt doing:

```python
reviews.groupby('points').points.count()
```

```python
#lets get the cheapest wine in each point value
reviews.groupby('points').price.min()
```

Each group we generate as being a slice of our DataFrame containing only data with values that match.

The dataframe is accessible using **apply()** method, we can manipulate the data in any way we see fit.



```python
reviews.groupby('winery').apply(lambda df:df.title.iloc[0])
```

```python
#best wine by country and province
reviews.groupby(['country','province']).apply(lambda df:df.loc[df.points.idxmax()])
```

**agg()**

Method allows us to run a bunch of different funcions on your dataframe simultaneously.



```python
#simple statictical summary of the dataset as follows
reviews.groupby(['country']).price.agg([len,min,max])
```

```python
col_df=list(reviews.columns)

for i in col_df:
  print(i)
```

**Multi-indexes**

groupby() method sometimes results in what is called a multi-index.

A multi-index differs from regular index because it has multiple levels



```python
#lets get the length of description
countries_reviewed=reviews.groupby(['country','province']).description.agg([len])

countries_reviewed
```

```python
mi=countries_reviewed.index
type(mi)
```

**NOTE**: Often when dealing with multi-index methods we convert it back to a regular index, use **reset_index()** method.

```python
countries_reviewed.reset_index()
```

**Sorting**

To get the data the data inthe order we want it we can sort it ourselves. **sort_values()** mthod is handy for this.

```python
countries_reviewed=countries_reviewed.reset_index()

countries_reviewed.sort_values(by='len')
```

```python
#set it to sort in descending order
countries_reviewed.sort_values(by='len',ascending=False)
```

```python
#to sort by index let use the method by default
countries_reviewed.sort_index()
```

```python
#we can sort by more than one column
countries_reviewed.sort_values(by=['country','len'])
```

# **Group_by() method**

Its like saying **split my data into groups do something to each group, then combine the results**

**Split-Apply-combine**

You split the data into groups, apply an operation(like sum or mean), then combine the results back.



```python
#df
df=pd.DataFrame({
    "Category":['A','A','B','B','C'],
    "Value":[10,20,30,40,50]
})

#display
df
```

```python
#sum values by category
grouped_df=df.groupby('Category').Value.sum()
```

```python
#display
grouped_df
```

We are also able to perform other functions like **mean(), max(), min()**

```python
means_df=df.groupby('Category').Value.mean()

#display
means_df
```

We can use **agg()** method and perform multiple stats at once

```python
agg_result=df.groupby("Category").Value.agg(['sum','mean','max','min'])

#display
agg_result
```

You can group by more than one column at once

```python
df2 = pd.DataFrame({
    "Team": ["A", "A", "B", "B", "C", "C"],
    "Year": [2020, 2021, 2020, 2021, 2020, 2021],
    "Points": [10, 15, 20, 25, 30, 35]
})

df2
```

```python
#multiple columns groupby
multi_cols=df2.groupby(['Team','Year']).Points.sum()
```

```python
#display
multi_cols
```

```python
mi=multi_cols.index  #the result isa multi-level index(Team+Year)
type(mi)
```

We can also apply our own custom python functions using the **apply() method** as seen below:

```python
#subtract max value with the min value
def value_range(x):
  return x.max()-x.min()

range_result=df.groupby('Category').Value.apply(value_range)

#display
range_result
```

**Real-World Example (Wine-style 🍷)**

```python
wine_df = pd.DataFrame({
    "wine_type": ["red", "red", "white", "white", "red"],
    "points": [88, 92, 85, 90, 95]
})

wine_df
```

```python
#lets compute the average points per wine type
avg_points=wine_df.groupby('wine_type').points.mean()

#display
avg_points
```

**Sorting in pandas (sort_values & sort_index)**

You can sort a data in pandas either by values or by index

**sort_values()** sorts by columns

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Allan", "Mary", "John"],
    "Age": [25, 30, 20]
})

# Sort by Age ascending
df.sort_values("Age")
```

```python
#sort in descending order
df.sort_values("Age", ascending=False)
```

**sort_index()** sorts by row index


```python
df.sort_index()
```
