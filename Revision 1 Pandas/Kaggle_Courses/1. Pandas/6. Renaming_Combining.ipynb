# **Renaming and Combining**

**rename()** method lets you change index names or column names.

Example: Change points column name to score:

```python
reviews.rename(columns={"points":"score"})
```

```python
#another method- Rename the series points
reviews['points'].rename("score")
```

```python
#Lets rename an index
reviews.rename(index={0: 'firstEntry',1:'secondEntry'})
```

**Combining**

When performing operations on a dataset, we will sometimes need to combine different DataFrames and/or Series in non-trivial ways.

**concat() method** can be used to combine two dataframes.

```python
#lets create two dataframes

df_1=pd.DataFrame({'Name': ['Alice', 'Bob', 'Charlie'],
        'Age': [25, 30, 35],
        'City': ['New York', 'London', 'Paris']}
)

df_1

```

```python
#create the second one
df_2=pd.DataFrame({
    'Cars':['Mac','Ferari','Toyota'],
    'Price':[10000,120300,20400],
    'Year':[2020,2000,1995]
})

df_2
```

```python
#now merge the dataframes
pd.concat([df_1,df_2])
```

**join()**method lets you combine DataFrame objects with a common index.



```python
#this is a left joint and uses index to merge the dataframes
result_df=df_1.join(df_2)

result_df
```
