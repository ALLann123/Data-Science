# **Maps**

A functions that takes one set of values and "maps" them to another set of values.

map() -Example we wanted to remean the scores the wines received to 0:

```python
reviews_points_mean=reviews['points'].mean()

reviews.points.map(lambda p:p - reviews_points_mean)

#we get a new series from our function above
```

**apply()**

Method is used when we want to transform a whole dataframe by calling a custom method on each row
.

```python
def remean_points(row):
  row.points=row.points-reviews_points_mean
  return row

#apply the above method to all the columns
reviews.apply(remean_points,axis='columns')
```

```python
#EXAMPLE: Combining country and region information inthe dataset
reviews['country']+"-"+reviews['region_1']
```

**NOTE**: The above operators are faster tat map() and apply() because they use speed ups built into pandas. However, map() and apply() can do more advanced things like applying conditional logic.

**Map()**

Is very handy for applying operations to each element of a series.

*NOTE*: series is a column in a dataframe.


```python
#Example- Mapping Series

#create a series of numbers
s=pd.Series([1,2,3,4,5,6,7])

#square each element inthe column
squared_s=s.map(lambda x:x**2)

#display
squared_s
```

```python
#Mapping with dictionary
#series of fruit
fruits=pd.Series(['apple','banana','mango','apple','banana'])

#map each fruit to a color
color_map={'apple':'red','banana':'yellow','mango':'orange'}

#run .map() method for the operation
fruit_colors=fruits.map(color_map)

#display
fruit_colors
```

```python
#Mapping with a function
names=pd.Series(['Allan','Mary','Alvin'])

#convert to lowercase using map()
lowercase_names=names.map(str.lower)

#display
lowercase_names
```

**3. Clean text with functions**

Strip whitespace

Convert to lowercase

```python
name=pd.Series(['Alice  ','  Bob','Charlie'])

#remove white spaces and convert to lowercase
cleaned_name=name.map(lambda x:x.strip().lower())

#display
cleaned_name
```

**4. Apply Multiple Conditions**

You have numbers: [10, 25, 30, 45]. Map each number to:

"Small" if < 20

"Medium" if 20–40

"Large" if > 40

```python
#create series
numbers=pd.Series([10,25,30,45])

#display
numbers
```

```python
def num_descip(nums):
  if nums>40:
    return "Large"
  elif nums>=20:
    return "Medium"
  elif nums<20:
    return "Small"

#map() method
manipulate_nums=numbers.map(num_descip)
```

```python
#display
manipulate_nums
```
