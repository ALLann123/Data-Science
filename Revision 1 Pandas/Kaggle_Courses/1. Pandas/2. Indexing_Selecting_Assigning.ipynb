# **Indexing in Pandas**

"reviews['country'][0]" this method works nicely just like they do inthe rest of the python ecosystem.

Pandas has its own access operators, **loc**, **iloc**. Used for advanced operations

**Index-based Selection**

Refers to selecting data based on its own numerical position in the data. iloc follows this.



```python
#use iloc to select the first row
reviews.iloc[0]
```

**NOTE**: Both *loc* and *iloc* are row-first, column-second. This is the opposite of what we have in natie python, which is column first then row second

```python
#getting a column('country') using iloc
reviews.iloc[:,0]
```

The **: operator**, comes from native python which means select "everything". Inthis case indicates select a range of values.

```python
#select the first 4 rows inthe column country
reviews.iloc[:4,0]
```

```python
#select the second and third row inthe column
reviews.iloc[1:3,0]
```

```python
#we are able to select by passing a list of indexes
reviews.iloc[[1,2,3,5,6,66,234],0]
```

**Negative numbers** can be used in selection. Will start counting from the bottom of the rows.

```python
#select last 5 rows
reviews.iloc[-5:]
```

# **Label-Based Selection**

loc operator is a label based operator. In this its the data index value thats important and not its position that matters.


```python
#select the first row and column value
reviews.loc[0,'country']
```

```python
#select columns
reviews.loc[:, ['taster_name', 'taster_twitter_handle', 'points']]
```

# **Manipulation the index**

Label-based(loc) selection derives its power from the labels in the index.

The index can be manipulated any way we see fit

**set_index()** method can be used to do the job. We can specify any column to be index

```python
#set title to be the index
reviews.set_index('title')
```

**Conditional Selection**

Doing intresting things with the data we have to ask questions based on conditions.




```python
#get wines from italy
reviews.loc[reviews.country=="Italy"]
```

```python
#lets get wines in italy better than the average point
reviews.loc[(reviews.country=="Italy")& (reviews.points>=90)]
```

```python
#wine made in italy. and or rated below above
reviews.loc[(reviews.country=="Italy") | (reviews.points>=90)]
```

**isin()** method is another built-in conditional selector. Lets you select data in a list of values.

```python
#select wines from italy and france
reviews.loc[reviews.country.isin(['Italy','France'])]
```

**isnull() method** lets you highlight values which are empty.

**notnull() method**  lets you select values which are not empty

```python
#to filter out wines lacking a price tag in a dataset

reviews.loc[reviews.price.notnull()]
```
