# **Data Types and Missing Values**

**dtype** refers to the data type of a column in a DataFrame or Series.

To get the dtype property of a specific column use the **dtype** attribute adter the column in the reviews


```python
#get the dytpe of the price column
reviews['price'].dtype
```

```python
#for the whole DataFrame
reviews.dtypes
```

Data types tell us something about how pandas is storing the data internally.

*NOTE*: Columns consisting entirely of strings do not get their own type; they are given the object type

Column data type conversion is also possible whenever such a conversion makes sense by using **astype()** function.

```python
#transform points column from int64 to float64
reviews['points'].astype('float64')
```

```python
#A dataFrame or Series index has its own dtype, too:
reviews.index.dtype
```

**Missing Data**

Missing values are represented by **NaN** short for "Not a Number". Null values are of float64 dtype

To select null entries in pandas use **pd.isnull()**

*NOTE*:To select non null use **pd.notnull()**

```python
#get null values inthe country column
reviews[pd.isnull(reviews.country)]
```

```python
#Another way to do the previous cell
reviews[reviews['country'].isnull()]
```

```python
#Identify number of missing null values in the whole dataframe per column
reviews.isnull().sum()
```

Replace missing values in pandas using **fillna()**/**fillna()** methods

```python
#replace null values inthe region_2 column
reviews['region_2'].fillna("Unknown")
```

We can use **replace()** method to replace values in a column.

The above method is handy in replacing missing data which is given some kind of sentinel value inthe dataset: things such as "Unknown, Invalid"

EG: Twitter handle from @kerinokeefe changed to @kerino

```python
reviews['taster_twitter_handle'].replace("@kerinokeefe", "@kerino")
```
