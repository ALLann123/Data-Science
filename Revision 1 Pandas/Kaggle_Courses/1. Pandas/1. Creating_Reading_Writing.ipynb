# **Creating, Reading and Writing**

Pandas is a python library for data analysis.


```python
import pandas as pd
```

**Creating data**

Objects of pandas:
1. **DataFrame**- Is a table. Contains an array of individual entries, each having a certain value. (Made up series)

We use the pd.DataFrame() method to create a dataframe object


```python
#create an example data frame
pd.DataFrame({
    "Cars":['Ferari', 'Bugatti', 'Ford'],
    "Price":[100000,223000,89999],
    "Reviews":['Fancy','Expensive','Quick']
})
```

2. **Series** -is a sequence of data values. If a dataframe is a table then a series is a list.

*Note*: A series is a single column of a DataFrame.



```python
#create series
pd.Series([1,2,3,4,5])
```

```python
#we can assign index name to a series but it does not have a column name
pd.Series([23,24,25], index=['Mark','Jaba','Allan'])
```

# **Reading data Files**

Most of the time we work with data that already exists. Data can be stored in a number of formats, the most basic being CSV(Comma Separated Value) file.

*Note*: Use **pd.read_csv()**  function to read a csv file to a DataFrame.

Download the dataset:https://www.kaggle.com/datasets/christopheiv/winemagdata130k

```python
#load csv file
wine_reviews=pd.read_csv('sample_data/winemag-data.csv')
```

```python
#use head method to view the first rows of the df
wine_reviews.head()
```

```python
#Shape attribut displays how large the resulting dataframe is
wine_reviews.shape  #(rows, columns)
```

```python
#Lets see the column names
list(wine_reviews.columns)
```

```python
#get info- column data types, misssing values can be seen from this
wine_reviews.info()
```

```python
#Find columns with missing values and number missing in the columns
wine_reviews.isnull().sum()
```

```python
#lets see countries we got reviews from
wine_reviews['country'].value_counts()
```

# **Saving Dataframe to File**

We are also able to save a dataframe to different file formats for this example lets go with csv file

```python
#create a dataframe
animals = pd.DataFrame({'Cows': [12, 20], 'Goats': [22, 19]}, index=['Year 1', 'Year 2'])
animals
```

```python
#save it to csv
animals.to_csv('animals.csv')
```

```python
#lets clean our data we loaded previously
#--Drop null values
wine_reviews_new=wine_reviews.dropna()
```

```python
#now lets select relevant columns
wine_reviews_new=wine_reviews_new[['country','description','taster_twitter_handle','winery']]
```

```python
#Save it now
wine_reviews_new.to_csv('wine_reviews_cleaned.csv')
```
